### Streaming And Batch

### Streaming
Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.
Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence, short answer")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence, short answer"):
    print(chunk.text, end="|", flush=True)
```

```python
model.invoke("Why do parrots have colorful feathers?short answer, short answer")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers? short answer"):
    print(chunk.text, end="|", flush=True)
```


#### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = model.batch([
    "Why do parrots have colorful feathers? short answer",
    "How do airplanes fly? short answer",
    "What is quantum computing? short answer"
])
for response in responses:
    print(response)
```

```python
model.batch(
    ["Why do parrots have colorful feathers?short answer, short answer",
    "How do airplanes fly? short answer",
    "What is quantum computing? short answer"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
```
