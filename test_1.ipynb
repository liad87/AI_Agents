```python
#import ollama
#for running on CPU only run this command on CMD: set OLLAMA_NUM_GPU=0
from langchain_ollama import ChatOllama
```

```python

# אתחול המודל המקומי באמצעות החבילה החדשה והרשמית
#llm = ChatOllama(model="qwen2.5:3b", temperature=0)
llm= ChatOllama(model="tinyllama", temperature=0)

# בדיקה שהמודל מגיב בתוך ה-Jupyter
try:
    response = llm.invoke("תגיד 'הכל עובד בהצלחה!'")
    print("🤖 תשובת המודל:")
    print(response.content)
except Exception as e:
    print(f"❌ עדיין יש שגיאה: {e}")
```
